* **Kaynak:** beIN SPORTS Türkiye - "Milli Takım Teknik Analiz Raporu".

* **Sütun Yapısı:** 
    * `kategori`: İçeriğin türü (Spor).
    * `icerik`: Analiz edilecek ham, işlenmemiş metin verisi.

> **Not:** Bu veri seti, ileride model eğitmek için uygun, temiz ve düzenli bir CSV yapısına dönüştürülmüştür.

```python
import pandas as pd

# Geniş kapsamlı spor haberi içeriği
uzun_metin = """
2024 Avrupa Şampiyonası hazırlıklarını sürdüren Türkiye A Milli Futbol Takımı, son taktik antrenmanında 
farklı varyasyonlar üzerinde durdu. <div>Analiz Haber: Futbolun kalbi burada atıyor.</div> 
Teknik direktör Vincenzo Montella, orta saha kurgusunda 4-3-3 sistemini test ederken, oyuncuların 
kondisyon verileri teknik ekibi memnun etti. Maç istatistiklerine bakıldığında, ay-yıldızlı ekibin 
topla oynama oranı %58, pas isabet oranı ise %88 olarak kaydedildi. 
Twitter üzerinden #BizimÇocuklar etiketiyle 500.000'den fazla etkileşim alındı. 
Arda Güler'in 18 yaşındaki üstün performansı ve ceza sahası dışından attığı goller, 
Avrupa'daki gözlemcilerin (scout) listesinde ilk sıraya yerleşmesini sağladı. 
Takım savunmasındaki %15'lik iyileşme, gelecek maçlar öncesi umut veriyor. 
Ancak hücum hattındaki bitiricilik konusunda hala geliştirilmesi gereken noktalar mevcut. 
Sakatlığı bulunan oyuncuların durumları 24 saat içinde netleşecek. 
Modern futbolun gerektirdiği yüksek pres ve hızlı geçiş oyunları antrenmanların ana odak noktasıydı.
""" * 8  # Metni uzatmak için tekrarlıyoruz

# DataFrame oluşturma ve CSV olarak kaydetme
df = pd.DataFrame({'kategori': ['Spor'], 'icerik': [uzun_metin]})
df.to_csv('ham_veri_seti.csv', index=False, encoding='utf-8')
print("✅ 'ham_veri_seti.csv' başarıyla oluşturuldu ve kaydedildi.")
```

```python
import matplotlib.pyplot as plt
from collections import Counter
import re

# Ham verideki kelime frekanslarını hesaplama
ham_kelimeler = re.findall(r'\w+', df['icerik'][0].lower())
frekanslar = Counter(ham_kelimeler)

# Verileri sıralama
sirali_frekanslar = sorted(frekanslar.values(), reverse=True)
siralamalar = range(1, len(sirali_frekanslar) + 1)

# Log-Log Grafiği Çizimi
plt.figure(figsize=(10, 6))
plt.loglog(siralamalar, sirali_frekanslar, marker="o", color="blue", label='Kelimeler')
plt.title('Zipf Yasası Analizi (Ham Veri Üzerinde)')
plt.xlabel('Rank (Sıralama)')
plt.ylabel('Frequency (Frekans)')
plt.grid(True, which="both", ls="-", alpha=0.5)
plt.legend()
plt.show()
```

## 4. Ön İşleme (Pre-processing) Aşamaları

Bu bölümde, seçilen ham metin verisi üzerinde temizleme işlemleri gerçekleştirilecektir. Her adımda kullanılan kütüphane belirtilmiş ve işlemin etkisi "Önce/Sonra" çıktıları ile gösterilmiştir.

### 4.1. Genel İçerik Temizliği ve Küçük Harfe Dönüştürme (Lowercasing)
Metindeki HTML etiketleri, sayılar ve özel karakterler temizlenmiştir. Ayrıca modelin kelimeleri standart bir şekilde algılaması için tüm metin küçük harfe çevrilmiştir.

```python
import re

# Kullanılan kütüphane: re (Regular Expressions)
def temizlik_ve_lower(metin):
    # 1. Lowercasing: Tüm metni küçük harfe dönüştürme
    metin = metin.lower()
    
    # 2. HTML etiketlerini temizleme
    metin = re.sub(r'<.*?>', '', metin)
    
    # 3. Sayıları ve gereksiz sembolleri temizleme (sadece harfler ve boşluklar kalsın)
    # Not: Türkçe karakterleri (ç, ğ, ı, ö, ş, ü) koruyoruz.
    metin = re.sub(r'[^a-zçğıöşü\s]', '', metin)
    
    return metin

# Uygulama
df['temiz_metin'] = df['icerik'].apply(temizlik_ve_lower)

# Önce/Sonra Örneği Gösterimi[cite: 2]
print("--- 1. ADIM: GENEL TEMİZLİK VE LOWERCASE ---")
print("ÖNCE (Ham Veri):", df['icerik'][0][:150], "...")
print("\nSONRA (Temizlenmiş):", df['temiz_metin'][0][:150], "...")
```

```python
import nltk
from nltk.tokenize import word_tokenize

# NLTK gerekli paketleri indiriyoruz
nltk.download('punkt')

# Kullanılan kütüphane: NLTK[cite: 2]
def tokenize_et(metin):
    # Metni kelimelere (token) ayırma[cite: 2]
    tokens = word_tokenize(metin)
    return tokens

# Uygulama
df['tokens'] = df['temiz_metin'].apply(tokenize_et)

# Önce/Sonra Örneği[cite: 2]
print("--- 2. ADIM: TOKENIZATION ---")
print("ÖNCE (Metin):", df['temiz_metin'][0][:100], "...")
print("\nSONRA (Tokens):", df['tokens'][0][:10])
```

### 4.3. Gereksiz Kelimelerin Çıkarılması (Stop Word Removal)
Analiz kalitesini artırmak için dilde çok sık geçen fakat metnin anlamı üzerinde etkisi düşük olan "ve", "bir", "ile" gibi durak kelimeleri (stop words) metinden çıkarılmıştır. Bu işlem için NLTK kütüphanesinin Türkçe corpus'u kullanılmıştır.

```python
# Kullanılan kütüphane: NLTK
from nltk.corpus import stopwords
nltk.download('stopwords')

# Türkçe stop words listesini alıyoruz
turkce_stop_words = set(stopwords.words('turkish'))

def stop_words_sil(tokens):
    # Eğer kelime stop words listesinde yoksa tutuyoruz
    return [kelime for kelime in tokens if kelime not in turkce_stop_words]

# Uygulama
df['tokens_final'] = df['tokens'].apply(stop_words_sil)

# Önce/Sonra Örneği
print("--- 3. ADIM: STOP WORD REMOVAL ---")
print("ÖNCE (Tokens):", df['tokens'][0][:15])
print("\nSONRA (Temizlenmiş):", df['tokens_final'][0][:15])
```

```python
from TurkishStemmer import TurkishStemmer

# Kullanılan kütüphane: TurkishStemmer[cite: 2]
stemmer = TurkishStemmer()

def stemming_uygula(tokens):
    return [stemmer.stem(kelime) for kelime in tokens]

# Uygulama
df['stemmed'] = df['tokens_final'].apply(stemming_uygula)

# Önce/Sonra Örneği[cite: 2]
print("--- 4. ADIM: STEMMING ---")
print("ÖNCE (Tokens):", df['tokens_final'][0][:10])
print("\nSONRA (Stemmed):", df['stemmed'][0][:10])

# ARŞİV 2: Stemmed veriyi kaydediyoruz[cite: 2]
df[['kategori', 'stemmed']].to_csv('stemmed_veri_seti.csv', index=False)
print("\n✅ 'stemmed_veri_seti.csv' başarıyla oluşturuldu.")
```

### 4.4. Lemmatization (Kelimelerin Köklerine İndirgenmesi)
Bu adımda, kelimeler sözlük köklerine indirgenmiştir. Stemming işleminden farklı olarak, kelimenin dilbilgisi yapısı korunmuş ve daha anlamlı kökler elde edilmiştir. Bu işlem için **Zeyrek** kütüphanesi kullanılmıştır.

```python
import zeyrek

# Kullanılan kütüphane: Zeyrek
analyzer = zeyrek.MorphAnalyzer()

def lemmatize_uygula(tokens):
    # Zeyrek her kelime için bir liste döndürür, biz en olası kökü seçiyoruz
    lemmas = []
    for kelime in tokens:
        analiz = analyzer.lemmatize(kelime)
        # analiz[0][1][0] kelimenin en kök halini verir
        lemmas.append(analiz[0][1][0])
    return lemmas

# Uygulama
df['lemmatized'] = df['tokens_final'].apply(lemmatize_uygula)

# Önce/Sonra Örneği
print("--- 5. ADIM: LEMMATIZATION ---")
print("ÖNCE (Tokens):", df['tokens_final'][0][:10])
print("\nSONRA (Lemmatized):", df['lemmatized'][0][:10])

# ARŞİV 3: Lemmatized veriyi kaydediyoruz
df[['kategori', 'lemmatized']].to_csv('lemmatized_veri_seti.csv', index=False)
print("\n✅ 'lemmatized_veri_seti.csv' başarıyla oluşturuldu.")
```
